# K-Nearest Neighbors (K-NN)

## Main Math for K-Nearest Neighbors Classification

K-Nearest Neighbors (K-NN) classification predicts a class using the labels of nearby training observations.

In this notebook:

- $x$: input vector containing age and estimated salary.
- $y$: actual class label, where 0 means not purchased and 1 means purchased.
- $\hat{y}$: predicted class label.
- $n$: number of training observations.
- $k$: number of nearest neighbors used for prediction; this notebook uses $k=5$.

### 1. The classification equation

K-NN with uniform weights predicts the class receiving the most votes among the nearest neighbors:

$$
\hat{y}(x)
=
\arg\max_{c\in\mathcal{C}}
\sum_{i\in N_k(x)}\mathbf{1}(y_i=c)
$$

- **$\mathcal{C}$:** set of possible classes; here, $\mathcal{C}=\{0,1\}$.
- **$N_k(x)$:** indices of the $k$ training observations closest to $x$.
- **$\mathbf{1}(y_i=c)$:** equals 1 when neighbor $i$ belongs to class $c$, and 0 otherwise.
- **$\arg\max$:** selects the class with the largest vote count.

For example, if the five nearest neighbors have labels $0,1,1,0,1$, class 1 receives three votes:

$$
\hat{y}(x)=1
$$

For binary classification with uniform weights, an odd $k$ avoids a tie in the class vote.

### 2. Prediction errors

Class labels represent categories, so classification is evaluated by whether the predicted label matches the actual label.

For observation $i$, define the error indicator:

$$
\ell_i=\mathbf{1}(y_i\neq\hat{y}_i)
$$

It equals 0 for a correct prediction and 1 for an incorrect prediction.

For an evaluation set containing $N$ observations:

$$
\mathrm{Error\ Rate}
=
\frac{1}{N}\sum_{i=1}^{N}\mathbf{1}(y_i\neq\hat{y}_i)
$$

$$
\mathrm{Accuracy}
=
\frac{1}{N}\sum_{i=1}^{N}\mathbf{1}(y_i=\hat{y}_i)
=
1-\mathrm{Error\ Rate}
$$

The notebook also uses a confusion matrix. With class order $[0,1]$, rows representing actual classes, and columns representing predicted classes:

$$
\mathrm{Confusion\ Matrix}
=
\begin{bmatrix}
TN & FP\\
FN & TP
\end{bmatrix}
$$

Here, $TN$ and $TP$ count correct predictions of classes 0 and 1; $FP$ counts actual 0s predicted as 1, and $FN$ counts actual 1s predicted as 0.

### 3. Finding the nearest neighbors

The notebook uses Euclidean distance, equivalent to Minkowski distance with exponent 2.

For two standardized input features:

$$
d(z,z_i)
=
\sqrt{(z_1-z_{i1})^2+(z_2-z_{i2})^2}
$$

Here, $z$ is the standardized input being predicted, and $z_i$ is standardized training observation $i$.

The model selects the $k$ observations with the smallest distances.

Because age and salary have different scales, each feature is standardized before calculating distances:

$$
z_j=\frac{x_j-\mu_j}{\sigma_j}
$$

Here, $\mu_j$ and $\sigma_j$ are the mean and standard deviation of feature $j$, calculated from the training data. This equation assumes $\sigma_j>0$.

The same training-derived transformation is applied to test observations and new inputs. This prevents salary's larger numerical scale from dominating the distance calculation simply because of its units.

### 4. Calculating class probabilities and choosing $k$

With uniform weights, the estimated probability of class $c$ is the fraction of neighbors belonging to that class:

$$
\hat{P}(y=c\mid x)
=
\frac{1}{k}
\sum_{i\in N_k(x)}\mathbf{1}(y_i=c)
$$

The predicted class has the largest estimated probability:

$$
\hat{y}(x)
=
\arg\max_{c\in\mathcal{C}}\hat{P}(y=c\mid x)
$$

For the neighbor labels $0,1,1,0,1$:

$$
\hat{P}(y=1\mid x)=\frac{3}{5}=0.6,
\qquad
\hat{P}(y=0\mid x)=\frac{2}{5}=0.4
$$

Therefore, the predicted class is 1. These fractions describe the local training neighborhood; they are not guarantees about an individual outcome.

The choice of $k$ controls flexibility:

- **Small $k$:** predictions respond strongly to local detail and noise.
- **Large $k$:** predictions use broader neighborhoods and may miss local class patterns.

A candidate value can be selected by minimizing validation error:

$$
k^*
=
\arg\min_{k\in\mathcal{K}}
\mathrm{Error\ Rate}_{\mathrm{validation}}(k)
$$

Here, $\mathcal{K}$ is the set of candidate values. Cross-validation can also be used, with feature scaling fitted only on the training portion of each split.

K-NN does not fit global slopes or intercepts. It retains the training observations and predicts each new class from its neighbors.

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv('Social_Network_Ads.csv')
X=  dataset.iloc[:, :-1].values
y= dataset.iloc[:, -1].values
```

## Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

```python
print(X_train)
```

```python
print(y_train)
```

```python
print(X_test)
```

```python
print(y_test)
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

## Training the K-NN model on the Training set

```python
from sklearn.neighbors import KNeighborsClassifier
classifier = KNeighborsClassifier(n_neighbors = 5, metric = 'minkowski', p = 2)
classifier.fit(X_train, y_train)
```

## Predicting a new result

```python
print(classifier.predict(sc.transform([[30, 87000]])))
```

## Predicting the Test set results

```python
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

## Making the Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```

## Visualising the Training set results

```python
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_train), y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 1),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 1))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('K-NN (Training set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
```

## Visualising the Test set results

```python
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_test), y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 1),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 1))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('K-NN (Test set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
```
